## Data Extraction - Pandas

```python
import pandas as pd
import numpy as np
```

The `pd.read_csv()` function allows to read a CSV file and load its contents as a Pandas DataFrame object.

```python
# Read the dataset, but load just 2000 rows from the CSV file.
dataset = pd.read_csv("./steam_reviews.csv", header="infer", nrows=2000)

# Explore the extracted rows.
dataset.shape
dataset.columns

# It is possible to get a concise summary of the dataset's contents using the info() function.
dataset.info()
```

After exploring a dataset, it is possible to pre-process data to make it ready for large-scale usage.

```python
def dateparse(time_in_secs):
  # A function to parse dates.
  return pd.to_datetime(time_in_secs, unit="s")

# Apply the date parse when loading the dataset.
dataset = pd.read_csv("./steam_reviews.csv",
                      header="infer",
                      nrows=2000,
                      parse_dates=["timestamp_created", "timestamp_updated", "author.last_played"],
                      date_parser=dateparse)
dataset.info()
```

```python
# Explore some functions and attributes associated to the datatime64[ns] datatype.
dataset.timestamp_created.min()
dataset.timestamp_created.max()
dataset.timestamp_created.dt.dayofweek
```

### Accessing Data

The simplest way to access some rows in a dataset is by using the `loc` function or via index slicing.

```python
dataset.loc[26] # Access row 26.
dataset[:3] # Access the first 3 rows.
```

Alternatively, it is possible to extract the contents of a dataset using conditional indexing, which consists of taking just the rows that satisfy a given condition.

```python
# Determine the languages of the reviews in the dataset.
dataset.language.unique()
dataset.language.nunique()

# Now, extract just the reviews written in English.
english_reviews = dataset[dataset.language == "english"]
len(english_reviews)
```

### Data Aggregation

The `groupby()` function allows to aggregate the contents of a dataset into groups that are based on the values taken by one or more columns.

```python
# Aggregate data by review language to see how many reviews have been written in a certain language.
dataset.groupby("language").review_id.count()

# Aggregate data by more attributes.
dataset.groupby(["language", dataset.timestamp_created.dt.hour]).review_id.count()
```

Additionally, the `groupby()` function can be used to generate descriptive statistics through the `describe()` method, eventually adding the `unstack()` method to format these statistics as a pivot table.

```python
dataset.groupby([dataset.timestamp_created.dt.hour, dataset.language]).describe()
dataset.groupby([dataset.timestamp_created.dt.hour, dataset.language]).describe().unstack()
```

## Data Visualization - Matplotlib

```python
# A simple plot of the number of reviews in a given language.
dataset["language"].value_counts().plot()
```

```python
# Alternatively, use a bar plot to represent the number of reviews in a given language.
dataset["language"].value_counts().plot.bar()
```

A good plot typically features a title and axis labels, which allow to understand the information shown by the plot, and an appropriate scale to properly visualize each value.

| Command | Plot |
| :---: | :---: |
| `hist` | Hisotgram |
| `box` | Boxplot |
| `kde` | Density Plot |
| `pie` | Pie Chart |
| `bar` | Bar Plot |

Additionally, it is possible to also add annotations or legends that, while not necessary, can help to better understand the contents of a plot.

```python
dataset["language"].value_counts().plot.bar(figsize=(18,7),
                                            title="Steam Reviews By Language (2021)",
                                            xlabel="Language",
                                            ylabel="Number of Reviews",
                                            color="#6BD19F")
```

`Matplotlib` is a popular library for data plotting and visualization.

```python
import matplotlib.pyplot as plt
```

```python
# Create a pie chart using Matplotlib.
plt.rcParams["figure.figsize"] = (8, 8)
plt.pie(dataset["language"].value_counts(),
        labels = dataset.language.unique(),
        shadow = True,
        autopct = "%.1f%%")
plt.title("Steam Reviews by Language (2021)", fontsize=20)
plt.axis("off")
plt.show()
```

```python
# Check the languages for which less than 10 reviews have been written.
data = dataset.language.value_counts()
labels = dataset.language.unique()

irrelevant_counts = data < 10
new_data = data[~irrelevant_counts]
new_data["other"] = data[irrelevant_counts].sum()
new_data

plt.rcParams["figure.figsize"] = (8, 8)
plt.pie(new_data,
        labels = new_data.index,
        shadow = True,
        autopct = "%.1f%%")
plt.title("Steam Reviews by Language (2021)", fontsize=20)
plt.axis("off")
plt.show()
```

### Plot Customization

It is also possible to customize plot axes using the `plt.xticks()` and `plt.yticks()` functions, which allow to set custom ticks on each axis, or the `plt.xlim()` `plt.ylim()` functions, which force value limits on each axis.

Additionally, one can improve plot visibility using the `plt.grid()` function, which adds a background grid to the plot, or the `plt.xscale()` and `plt.yscale()` functions, which allow to modify the value scale for each axis.

```python
hour_reviews = dataset.groupby(dataset.timestamp_created.dt.hour)

plt.figure()
hour_reviews.votes_helpful.sum().plot()
plt.xlabel("Hour")
plt.ylabel("Total Helpful Votes")
plt.xticks(range(0, 24))
plt.show()
```

### Creating Suplots

The `plt.subplot()` function allows to place different plots within the same image.

The image layout is structured as a grid that can be described using a three-digit number where:
- The first digit denotes the number of rows in the grid.
- The second digit denotes the number of columns in the grid.
- The third digit denotes the position of the subplot within the grid.
For instance, `plt.subplot(221)` is used to plot the first plot within a grid consisting of two rows and two columns.

```python
# Creating vertically aligned subplots.
plt.figure()
plt.subplot(211)
plt.plot([1, 2, 3])
plt.subplot(212)
plt.plot([4, 5, 6])
plt.show()
```
