## Defining Paths

For a network $G = (V, E)$, a path is a sequence of distinct nodes in which each node is adjacent to the next one.

Generally speaking, the length of a path is measured as the number of edges that compose it, although, for a weighted network, the sum of the edge weights is used instead.

In particular, the distance between two nodes corresponds to the length of the shortest path between them, with $d_{i, i} = 0 \text{ } \forall i \in V$ and $d_{i, j} = +\infty$ if no path connects $i$ to $j$.

However, it should be noticed that, for a directed graph, the distance from node $i$ to node $j$ might be not be equal to the distance from node $j$ to node $i$ as link directions must be respected.

From a broader perspective, distances in a network are measured using the breadth-first search algorithm, which acts as a variant of Dijkstra's algorithm for unweighted graphs.

In fact, this algorithm starts from a source node, which is at distance zero from itself, and finds all its neighbours, which are at distance one from the source node.

These neighbours are then added to a queue, which is used to iteratively compute all the distances in the graph until all the nodes have been visited by the algorithm.

```python
g_dist = ig.Graph(edges = [(0,1),(0,9),(0,13),(1,2),(2,3),(2,4),(2,5),
                           (5,6),(5,7),(5,8),(9,10),(9,11),(10,12),
                           (13,14),(13,15),(15,16),(15,18),(16,17),(18,19)],
                  directed = False)

fig, ax = plt.subplots(figsize=(5, 4))
ig.plot(
    g_dist,
    target=ax,
    vertex_label= [x.index for x in g_dist.vs()],
    vertex_label_color = "black",
    vertex_label_size = 10,
    vertex_size= 20,
    vertex_color= "grey",
    edge_color= "grey"
)
```

```python
# Compute the distances using Breadth-First Search.
distances = np.array(g_dist.distances(0,g_dist.vs())[0])
np.where(distances == 0, "red","grey")

# Plot distances.
fig, ax = plt.subplots(figsize=(5, 4))
ig.plot(
    g_dist,
    target=ax,
    vertex_label= g_dist.distances(0,g_dist.vs())[0],
    vertex_label_color = "black",
    vertex_size= 20,
    vertex_color=np.where(distances <= 1, "red","grey"),
    edge_color= "grey"
)


fig, ax = plt.subplots(2, 2, figsize=(7, 6))
ig.plot(
    g_dist,
    target=ax[0,0],
    vertex_label= g_dist.distances(0,g_dist.vs())[0],
    vertex_label_color = "black",
    vertex_size= 20,
    vertex_color=np.where(distances <= 1, "red","grey"),
    edge_color= "grey"
)
ig.plot(
    g_dist,
    target=ax[0,1],
    vertex_label= g_dist.distances(0,g_dist.vs())[0],
    vertex_label_color = "black",
    vertex_size= 20,
    vertex_color=np.where(distances <= 2, "red","grey"),
    edge_color= "grey"
)
ig.plot(
    g_dist,
    target=ax[1,0],
    vertex_label= g_dist.distances(0,g_dist.vs())[0],
    vertex_label_color = "black",
    vertex_size= 20,
    vertex_color=np.where(distances <= 3, "red","grey"),
    edge_color= "grey"
)
ig.plot(
    g_dist,
    target=ax[1,1],
    vertex_label= g_dist.distances(0,g_dist.vs())[0],
    vertex_label_color = "black",
    vertex_size= 20,
    vertex_color=np.where(distances <= 4, "red","grey"),
    edge_color= "grey"
)
```

### Average Path Lengths

If $G = (V, E)$ is an undirected graph, the average distance for a path is given by
$$\langle d \rangle = \frac{1}{\binom{|V|}{2}}\sum_{i \neq j} d_{i, j}$$
By contrast, if $G = (V, E)$ is a directed graph, the average distance for a path is defined as
$$\langle d \rangle = \frac{1}{|V|(|V| - 1)}\sum_{i \neq j} d_{i, j}$$
Overall, real networks tend to exhibit the "small-world" property because, due to the presence of hubs, any node can be reached from any other node in a small number of steps, resulting in $\langle d \rangle \propto \log(|V|)$.

### Connectedness in Networks

Sometimes, a network simply consists of multiple components, which represent subgraphs that are disconnected from each other.

In particular, it is possible to make a distinction between "weakly connected components", where each pair of vertices is connected only if edge direction is ignored, and "strongly connected components", where each pair of vertices is connected by a directed path.

```python
# Analyse components for the Zachary Karate Club network.
print(zkk.components().summary()) # Get an overview of the compoents.
print(zkk.components().sizes()) # Look at the size of each component.
print(zkk.components().membership) # See which nodes are associated to a certain component.
```

### Random Walks

On a graph $G = (V, E)$, a random walk is a stochastic process in which an agent starts from a source node and iteratively moves to a randomly chosen neighbour until a target node is reached.
